# Point operations

```python
from skimage import data
I = data.camera()
```

```python
f, axs = plt.subplots(2, 3, sharex='row', sharey='row')

axs[0, 0].imshow(I, cmap='gray', vmin=0, vmax=255)
axs[1, 0].hist(I.flatten(), 50)

axs[0, 1].imshow(I*1.2, cmap='gray', vmin=0, vmax=255)
axs[1, 1].hist((I*1.2).flatten(), 50)

axs[0, 2].imshow(I+50, cmap='gray', vmin=0, vmax=255)
axs[1, 2].hist((I+50).flatten(), 50)

for ax in axs[0, :]:
    ax.axis('off')
for ax in axs[1, :]:
    ax.grid(True)
```
