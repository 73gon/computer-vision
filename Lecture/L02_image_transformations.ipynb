```python
import numpy as np
import matplotlib.pyplot as plt
```

# Image coordinate transformations

```python
# Create a trivial image
w = 100
h = 80
I = np.zeros((h, w))
I[40:70, 30:50] = 1
```

```python
# Which form and what color?
plt.imshow(I,cmap='gray')
```

# Question
- Why do we see colors despite creating grayscale images?

```python
# How to transform the image now?
```

```python
# 1) build a target image to copy pixels to
Is = np.zeros_like(I)
```

```python
# 2) determine the pixel locations in the input image - 
#    create homogeneous coordinates right away
YXh = np.array([(y, x, 1) for x in range(w) for y in range(h)]).T
YXh
```

```python
# 3) convert to target pixel locations
tx, ty = 20, 15
theta = np.pi/8
A = np.array([
    [np.cos(theta), -np.sin(theta), tx],
    [np.sin(theta), np.cos(theta), ty],
    [0, 0, 1]
])

YXhs = np.dot(A, YXh)
YXhs
```

```python
# 4) Convert to discrete integer indices. Let's round for the moment.
YXhs = (YXhs+.5).astype('int') # int(x+0.5)≈round(x)
YXhs
```

```python
# 5) clip the coordinates
inside = np.logical_and.reduce([
    YXhs[0, :] >= 0,
    YXhs[0, :] < h,
    YXhs[1, :] >= 0,
    YXhs[1, :] < w,    
])
# shows how many transformed points remain in the image
np.count_nonzero(inside) / (w*h)
```

```python
# 6) copy pixels
Is[YXhs[0, inside], YXhs[1, inside]] = I[YXh[0, inside], YXh[1, inside]]
```

```python
f, (a1, a2) = plt.subplots(1, 2, sharex=True, sharey=True)
a1.imshow(I)
a2.imshow(Is)
```
